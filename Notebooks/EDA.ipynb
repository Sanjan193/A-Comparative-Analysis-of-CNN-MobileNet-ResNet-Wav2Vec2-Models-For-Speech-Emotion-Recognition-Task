# **Visualizing The Boxplots And Pair Plots Of The Features**

# **Imports**

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# loading the cleaned Data
# checking the dataset
data = pd.read_csv("/content/cleaned_dataframe.csv",
                   encoding='latin1')

data.head(5).T

```

# **Making the plots and Visualizing the data**

```python
# checking the entropy and enthalpy with respect to the number of elements
sns.set_style("whitegrid")
b = sns.FacetGrid(data, col="Num_of_Elem" )
b.map(sns.scatterplot, "dHmix", "dSmix", alpha=.7)
```

**now checking with the valence electron concentration , because it is proven to be useful to predict the state/phases an atom is...**

```python
# now checking with respect to the Phases
c = sns.FacetGrid(data, col="Phases" )
c.map(sns.scatterplot, "VEC", "dSmix", alpha=.7);
```

**trying to get the atom quantity with the state**

```python
c = sns.FacetGrid(data, col="Phases" )
c.map(sns.scatterplot,"Al", "Co", alpha=.7);
c.map(sns.scatterplot,"Al", "Cr", alpha=.7);
```

```python
sns.pairplot(data,hue='VEC',vars=['Num_of_Elem',
 'Density_calc',
 'dHmix',
 'dSmix',
 'dGmix',
 'Tm',
 'n.Para'])
```

# **Making a heatmap**
**seeing the correlations**

```python
temp_df = data[[
 # below are the columns that I want to see the correlation with
 'Num_of_Elem',
 'Density_calc',
 'dHmix',
 'dSmix',
 'dGmix',
 'Tm',
 'n.Para',
 'Atom.Size.Diff',
 'Elect.Diff',
 'VEC',
 'Homogenization_Temp',
 'Homogenization_Time',
 'Annealing_Time_(min)']]
corr_relation = temp_df.corr(numeric_only=True)
sns.heatmap(corr_relation)
plt.show()
```
